```python
import pandas as pd

# 1. Load kedua file CSV
# Ganti nama file sesuai dengan yang ada di foldermu
df_original = pd.read_csv(r'C:\Users\Daniel\Desktop\TechDesk\dataset\rcfa_100.csv')
df_new_cases = pd.read_csv(r'C:\Users\Daniel\Desktop\TechDesk\dataset\rcfa_150.csv')

print(f"Jumlah data awal: {len(df_original)}")
print(f"Jumlah data tambahan: {len(df_new_cases)}")

# 2. Menggabungkan kedua dataframe
# Kita gunakan ignore_index=True agar urutan nomor barisnya rapi kembali
df_combined = pd.concat([df_original, df_new_cases], ignore_index=True)

# 3. Opsional: Menghapus duplikat jika ada baris yang sama persis
df_combined = df_combined.drop_duplicates().reset_index(drop=True)

# 4. Cek hasil penggabungan
print(f"Total data setelah digabung: {len(df_combined)}")
print("\nBeberapa data teratas:")
print(df_combined.head())

# 5. Simpan ke file baru untuk di-upload ke LlamaFactory
df_combined.to_csv('rcfa_combined_dataset.csv', index=False)

print("\n✅ File 'rcfa_combined_dataset.csv' berhasil dibuat dan siap untuk training ulang!")
```

```python
import pandas as pd

df = pd.read_csv('rcfa_combined_dataset.csv')

# Ambil jumlah baris terbanyak sebagai target
target_count = df['failure_category'].value_counts().max()

# Fungsi untuk melakukan upsampling (menambah data yang kurang dengan cara menduplikasi secara acak)
df_balanced = df.groupby('failure_category').apply(lambda x: x.sample(target_count, replace=True)).reset_index(drop=True)

print("Distribusi setelah diseimbangkan:")
print(df_balanced['failure_category'].value_counts())

# Simpan hasil akhir
df_balanced.to_csv('rcfa_final_balanced.csv', index=False)
```
